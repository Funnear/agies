# 🎵 Jamendo API Exploration — EDM Track & Multi-Label Metadata

This notebook explores the Jamendo REST API (`/v3.0/tracks/`) to evaluate full-length track metadata, Creative Commons licensing details, micro-genre/mood tags, and MP3 audio preview links for EDM dataset assembly.

### Objectives
* Test connection authentication using `JAMENDO_CLIENT_ID` and standard helper utilities.
* Fetch raw JSON payloads targeting electronic/EDM genre tags (`tags=edm`).
* Inspect nested payload structures (`headers` and `results`).
* Evaluate core attributes (`id`, `name`, `artist_name`, `duration`, `audio`, `license_ccurl`) in a tabular pandas DataFrame.

```python
import os
import pandas as pd
from dotenv import load_dotenv
from notebooks.data_sources.api_helpers import fetch_raw_api_sample

load_dotenv()
print("Imports loaded successfully.")
```

## 1. API Configuration & Parameters

Set up base URLs, retrieve the Jamendo Client ID from local environment variables, and construct query parameters.

```python
# API Endpoint Configuration
JAMENDO_BASE_URL = "https://api.jamendo.com/v3.0/tracks/"
JAMENDO_CLIENT_ID = os.getenv("JAMENDO_CLIENT_ID", "709fa152")  # '709fa152' is Jamendo's public read test key

# Query Parameters
SEARCH_PARAMS = {
    "client_id": JAMENDO_CLIENT_ID,
    "format": "json",
    "limit": 5,
    "tags": "edm electronic",
    "audioformat": "mp32"
}

print("API configuration initialized.")
```

## 2. API Request Configuration & Raw Sample Fetch

Execute a test request against the `/v3.0/tracks/` endpoint to retrieve raw JSON records for target EDM tags.

```python
# Call generic fetch helper function
raw_payload = fetch_raw_api_sample(
    url=JAMENDO_BASE_URL,
    params=SEARCH_PARAMS
)

# Return raw data object directly to view raw JSON structure
raw_payload
```

## 3. Inspect Raw JSON Structure & Nested Payload Keys

Verify the root response schema (`headers` metadata) and inspect individual track records (`raw_payload['results']`).

```python
if raw_payload and "results" in raw_payload:
    records_list = raw_payload.get("results", [])
    print(f"Total records retrieved: {len(records_list)}")
    # Display first raw record structure
    display(records_list[0] if records_list else {})
```

## 4. Tabular Representation & Metadata Evaluation

Convert the dictionary array directly into a pandas DataFrame to verify presence of target fields (`id`, `name`, `artist_name`, `audio`).

```python
if raw_payload and "results" in raw_payload:
    df_jamendo = pd.DataFrame(raw_payload["results"])
    print("Returned Fields:", list(df_jamendo.columns))
    display(df_jamendo[["id", "name", "artist_name", "duration", "audio", "license_ccurl"]].head())
```
