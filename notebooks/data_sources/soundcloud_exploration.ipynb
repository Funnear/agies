# SoundCloud API Exploration — EDM Tracks & Metadata

This notebook explores track metadata fetched from the SoundCloud API (`https://api-v2.soundcloud.com/search/tracks`). It tests metadata fetching using search queries (targeting "edm"), inspects raw response structures, and evaluates key track attributes for downstream sub-genre analysis.

```python
import os
import requests
import pandas as pd
from dotenv import load_dotenv
from notebooks.data_sources.api_helpers import fetch_raw_api_sample


# Load environment variables
load_dotenv()
```

## 1. API Configuration & Parameters

Set up base URLs, retrieve authentication keys from local environment variables, and construct standard query parameters.

```python
CLIENT_ID = os.getenv("SOUNDCLOUD_CLIENT_ID")

if not CLIENT_ID:
    print("⚠️ SOUNDCLOUD_CLIENT_ID not set in .env file. Notebook will use fallback schema.")
else:
    print(f"✅ Loaded SoundCloud Client ID")
```

```python

# API Endpoint Configuration
SOUNDCLOUD_BASE_URL = "https://api-v2.soundcloud.com/search/tracks"
SOUNDCLOUD_CLIENT_ID = os.getenv("SOUNDCLOUD_CLIENT_ID")

# Query Parameters
SEARCH_PARAMS = {
    "q": "edm",
    "client_id": SOUNDCLOUD_CLIENT_ID,
    "limit": 15
}

print("API configuration initialized.")
```

## 2. Data Fetching & Payload Validation

Fetch raw track payloads using standard helper utilities and verify total available record counts.

```python

# Call generic fetch function with notebook constants
raw_payload = fetch_raw_api_sample(
    url=SOUNDCLOUD_BASE_URL,
    params=SEARCH_PARAMS
)

if raw_payload and "collection" in raw_payload:
    print(f"Successfully fetched raw payload. Items retrieved: {len(raw_payload.get('collection'))}")
else:
    print("Failed to fetch payload. Check logs/API key.")
```
