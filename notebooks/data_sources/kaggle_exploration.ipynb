## 2. Dataset 1: Discogs Masters (January 2025)

Download and inspect the primary Discogs master dataset dump.

```python
df_masters = download_kaggle_dataset(
    dataset_slug="ofurkancoban/discogs-datasets-january-2025",
    file_name="discogs_20250101_masters.csv"
)
eda_utils.print_eda_report(df_masters)
```

### Filtering & Style Analysis for Electronic Music

```python
# Count unique styles
display(eda_utils.unique_values_count(df_masters, columns=["styles_style"]))
# Filter for Electronic genre
electronic_df = df_masters[df_masters["genres_genre"] == "Electronic"]
print("Electronic Subset Shape:", electronic_df.shape)

# View unique styles list and preview sample rows
display(eda_utils.unique_values_list(electronic_df, "styles_style"))
display(electronic_df[['genres_genre', 'styles_style']].head(1))
```

```python
# Style distribution count
electronic_df['styles_style'].value_counts()
```
