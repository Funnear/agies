# Freesound API Exploration — EDM Audio Samples

This notebook explores audio sample metadata fetched from the [Freesound API](https://freesound.org/apiv2/). It tests sample fetching using search queries (targeting "edm"), inspects raw response structures, and evaluates key sample attributes for downstream analysis.

```python
import os
import pandas as pd
from dotenv import load_dotenv
from notebooks.data_sources.api_helpers import fetch_raw_api_sample

load_dotenv()
print("Imports loaded successfully.")
```

## 1. API Configuration & Parameters

Set up base URLs, retrieve authentication keys from local environment variables, and construct standard query parameters.

```python
# API Endpoint Configuration
FREESOUND_BASE_URL = "https://freesound.org/apiv2/search/text/"
FREESOUND_API_KEY = os.getenv("FREESOUND_API_KEY")

# Query Parameters
SEARCH_PARAMS = {
    "query": "edm",
    "token": FREESOUND_API_KEY,
    "page_size": 15
}

print("API configuration initialized.")
```

## 2. Data Fetching & Payload Validation

Fetch raw sample payloads using standard helper utilities and verify total available record counts.

```python
# Call generic fetch function with notebook constants
raw_payload = fetch_raw_api_sample(
    url=FREESOUND_BASE_URL,
    params=SEARCH_PARAMS
)

if raw_payload:
    print(f"Successfully fetched raw payload. Total count: {raw_payload.get('count')}")
else:
    print("Failed to fetch payload. Check logs/API key.")
```

## 3. Exploratory Data Analysis & Field Inspection

Convert raw JSON output to a pandas DataFrame to inspect available fields (`id`, `name`, `tags`, `license`, `username`) and initial records.

```python
if raw_payload and "results" in raw_payload:
    # Convert results array to DataFrame for quick inspection
    df_sample = pd.DataFrame(raw_payload["results"])
    print("Returned Fields:", list(df_sample.columns))
    display(df_sample.head())
```
