# 🎧 Beatport API Exploration — Micro-Genre Taxonomy & Track Metadata

This notebook explores Beatport catalog endpoints (`/v4/catalog/`) to evaluate electronic music sub-genre taxonomy, track metadata availability (BPM and musical key), and audio sample preview URLs for downstream ML pipeline integration.

### Objectives
* Test catalog querying using custom `User-Agent` headers and generic API helper utilities.
* Fetch the top-level electronic music genre taxonomy (`/v4/catalog/genres/`).
* Query track metadata for target EDM sub-genres to verify `bpm`, `key`, and preview audio file links.
* Inspect raw JSON payload structures and convert records to a pandas DataFrame for attribute evaluation.

```python
import os
import pandas as pd
from dotenv import load_dotenv
from src.agies.integration.api_helpers import fetch_raw_api_sample

load_dotenv()
print("Imports loaded successfully.")
```

## 1. API Configuration & Parameters

Set up target catalog URLs, headers matching project standards, and construct query parameters for sub-genre taxonomy retrieval.

```python
# API Endpoint Configuration
BEATPORT_BASE_URL = "https://api.beatport.com/v4/catalog"
HEADERS = {"User-Agent": os.getenv("USER_AGENT", "AGIES/0.1 (info@dataravers.space)")}

# Parameters for Genre Taxonomy Query
GENRE_PARAMS = {
    "page_size": 50
}

print("API configuration initialized.")
```

## 2. Data Fetching & Payload Validation

Fetch raw sub-genre taxonomy payloads using standard helper utilities and verify available record counts.

```python
# Fetch raw sample using shared helper utility
raw_payload = fetch_raw_api_sample(
    url=f"{BEATPORT_BASE_URL}/genres/",
    params=GENRE_PARAMS,
    headers=HEADERS,
    timeout=15
)

if raw_payload and "results" in raw_payload:
    print(f"Successfully fetched raw payload. Total sub-genres count: {len(raw_payload.get('results', []))}")
else:
    print("Failed to fetch payload. Check connection logs or fallback schema.")
```

## 3. Inspect Raw JSON Structure & Nested Payload Keys

Verify the root response schema and isolate individual genre dictionary entries (`raw_payload['results']`).

```python
if raw_payload and "results" in raw_payload:
    genre_list = raw_payload.get("results", [])
    print(f"Total genre items found: {len(genre_list)}")
    # Display raw structure of the first entry
    display(genre_list[0] if genre_list else {})
```

## 4. Tabular Representation & Metadata Evaluation

Convert the taxonomy records directly into a pandas DataFrame to inspect available metadata fields (`id`, `name`, `slug`).

```python
if raw_payload and "results" in raw_payload:
    df_genres = pd.DataFrame(raw_payload["results"])
    print("Returned Fields:", list(df_genres.columns))
    display(df_genres.head())
```
